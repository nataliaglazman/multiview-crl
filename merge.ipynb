```python
import pandas as pd
import numpy as np
```

```python
demog = pd.read_csv(
    "/Users/nataliaglazman/Downloads/ARC Builder Tables Oct 09 2026/1783_ID_Cohort_Analysis_PTDEMOG_09Oct2026.csv"
)
```

```python
labels = pd.read_csv("/Users/nataliaglazman/Desktop/PhD/projects/multiview-crl/data_local/labels_cleaned_3class.csv")
```

```python
labels
```

```python
demog = demog[["PTID", "PTGENDER", "PTRACCAT"]]
demog
```

```python
demog["Subject"] = demog["PTID"]
```

```python
# drop any duplicates in demog based on 'Subject'
demog = demog.drop_duplicates(subset=["Subject"])
```

```python
# merge using subject so that only the subjects in labels are kept and there are no duplicates
merged = pd.merge(labels, demog, on="Subject", how="left")
```

```python
merged = merged.drop(columns=["PTID"])
```

```python
merged.isna().sum()
```

```python
# count duplicates in merged by columns 'Subject' and 'PTGENDER' and 'PTRACCAT'

merged.duplicated(subset=["Subject"]).sum()
```
